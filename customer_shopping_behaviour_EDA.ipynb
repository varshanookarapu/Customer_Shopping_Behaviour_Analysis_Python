```python
#At first, my intuition was to replace the missing values with "Unknown", since there are only 37 missing values out of 3,900. However, after referring to the video, I realized that imputing the missing values
# using the median is a better approach. Another good point he made was regarding why we choose mean or median for imputation in the first place. In this case, we prefer the median over the mean because the mean can
#be significantly affected by outliers, which can skew the data distribution. Therefore, using the median provides a more robust estimate. Even with the median, he did not simply calculate the median of all the ratings
# and use that value to replace every missing entry. Instead, he calculated the median rating for each category and used the corresponding category-level median to fill in the missing values.
#This is a much more logical and meaningful way to impute the data because the typical rating can vary across different categories.
#Overall, I think this is a really good approach to handling the missing values in the dataset.
```

```python
#Based on the above observation , I will now try to find median review rating for every category.

# Another important distinction is median() reduces the data , transform("median") will keep the data set as is and then update the values

customer_behaviour_df.groupby("Category")["Review Rating"].median().reset_index(name="Median Review Rating")

```

```python
## Now i will replace the median review rating in all the missing values .

customer_behaviour_df["Review Rating"] = customer_behaviour_df["Review Rating"].fillna(customer_behaviour_df.groupby("Category")["Review Rating"].transform("median"))
```

```python
## Now I will check for duplicates and drop them  - bascially i am using duplicated function with sum to check and count the duplicates and if there are any duplicates i will use drop_duplicates() to remove them. and then put it in another df and check the shape of the data .
customer_behaviour_df.duplicated().sum()
## for now there's no duplicates in the data
```
